```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
POP_SIZE = 100
CHROM_SIZE = 10
GENERATIONS = 100
CROSS_RATE = 0.8
MUT_RATE = 0.01
X_MIN = -10
X_MAX = 10

def fitness(x):
    return 2*x**2 + 5*x

def decode(chrom):
    decimal = int("".join(map(str, chrom)), 2)
    return X_MIN + (decimal / (2**CHROM_SIZE - 1)) * (X_MAX - X_MIN)
```

```python
def hamming_diversity(pop):
    total = 0
    count = 0
    for i in range(len(pop)):
        for j in range(i+1, len(pop)):
            total += np.sum(pop[i] != pop[j])
            count += 1
    return total / count
```

```python
def run_ga(elite_rate):

    elite_size = int(POP_SIZE * elite_rate)

    pop = np.random.randint(2, size=(POP_SIZE, CHROM_SIZE))

    best_list = []
    mean_list = []
    div_list = []

    for g in range(GENERATIONS):

        decoded = np.array([decode(ind) for ind in pop])
        fitness_values = np.array([fitness(x) for x in decoded])

        best_list.append(np.min(fitness_values))
        mean_list.append(np.mean(fitness_values))
        div_list.append(hamming_diversity(pop))

        # Seleção por torneio
        new_pop = []

        elite_idx = fitness_values.argsort()[:elite_size]
        elites = pop[elite_idx]

        while len(new_pop) < POP_SIZE - elite_size:

            i1, i2 = random.sample(range(POP_SIZE), 2)
            parent1 = pop[i1] if fitness_values[i1] < fitness_values[i2] else pop[i2]

            i3, i4 = random.sample(range(POP_SIZE), 2)
            parent2 = pop[i3] if fitness_values[i3] < fitness_values[i4] else pop[i4]

            # Crossover
            if random.random() < CROSS_RATE:
                point = random.randint(1, CHROM_SIZE-1)
                child = np.concatenate((parent1[:point], parent2[point:]))
            else:
                child = parent1.copy()

            # Mutação
            for i in range(CHROM_SIZE):
                if random.random() < MUT_RATE:
                    child[i] = 1 - child[i]

            new_pop.append(child)

        pop = np.vstack((elites, np.array(new_pop)))

    return best_list, mean_list, div_list
```

```python
best, mean, div = run_ga(elite_rate=0.02)

print("Melhor fitness final:", best[-1])
print("Fitness médio final:", mean[-1])
print("Diversidade final:", div[-1])
```
